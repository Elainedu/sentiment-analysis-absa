### 將輸出結果儲存至.txt

```python
# 分類標籤
categories = ['整潔舒適情緒', '設施情緒', '服務情緒', '地點情緒', '性價比情緒', '其他情緒', '整體情緒']

# 情緒對應字典
emotion_mapping = {0: "無情緒", 1: "正面", 2: "中立", 3: "負面"}

# 創建一個空列表來存儲結果
output_results = []

#C:\Users\a0936\Downloads\w15-30-Alpaca-HumanAssistant短提示語(一問一答單輪對話)整理\w15-30-Alpaca-HumanAssistant短提示語(一問一答單輪對話)整理\情緒面相整理輸出格式.txt
# 迴圈遍歷每一列
for index, row_info in df.iterrows():
    result = f'以下這段文字提到那些面向?是正面還是負面的情緒?'
    for category in categories:
        if row_info[category] != 0:
            if '面向' in category:
                # 若分類面向的標籤為1，只印出面向標籤
                result += f'{category},'
            elif '情緒' in category:
                # 將情緒數值轉換為文字
                emotion_text = emotion_mapping.get(row_info[category], "未知情緒")
                result += f'{category}: {emotion_text} '
            else:
                result += f'{category}: {row_info[category]} '
    output_results.append(result)

# 將結果寫入檔案
output_file_path = 'C:\\Users\\USER\\OneDrive - 國立高雄科技大學\\桌面\\學姊練\\alpaca_gpt4_data_zh.json'
with open(output_file_path, 'w', encoding='utf-8') as output_file:
    for result in output_results:
        output_file.write(result + '\n')

print(f'結果已儲存至 {output_file_path}')
```
